```python
import pandas as pd
import matplotlib.pyplot as plt

# Load raw CSV
df_raw = pd.read_csv("gain vs directivity.csv", header=None)

# Remove completely empty columns
df_raw = df_raw.dropna(axis=1, how='all')

# Split the file into two parts
# First part: Directivity
directivity_data = df_raw.iloc[1:3, 0:2]
directivity_data.columns = ["Frequency", "Directivity"]
directivity_data = directivity_data.astype(float)

# Second part: Gain
gain_data = df_raw.iloc[4:6, 0:2]
gain_data.columns = ["Frequency", "Gain"]
gain_data = gain_data.astype(float)

# Merge both on Frequency
df = pd.merge(directivity_data, gain_data, on="Frequency")

# Plot
plt.figure()
plt.plot(df["Frequency"], df["Directivity"], linewidth=3,label="Directivity(dB)")
plt.plot(df["Frequency"], df["Gain"],linewidth=3, label="Gain(dB)")

plt.xlabel("Frequency(GHz)")
plt.title("Gain VS Directivity")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('Gain VS Directivity.png')


plt.show()
```
